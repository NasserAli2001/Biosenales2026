## 4. Representación de señales discretas

Para trabajar con señales discretas conviene guardar por separado los valores de la señal y los índices que les corresponden.

### Impulso unitario

````python
def impseq(n0, n1, n2):
    """Genera δ[n-n0] para n1 <= n <= n2. Devuelve (x, n)."""
    n = np.arange(n1, n2 + 1)
    x = (n == n0).astype(float)
    return x, n

x_imp, n_imp = impseq(0, -3, 3)
print("n =", n_imp)
print("δ[n] =", x_imp)
````

### p) Escalón unitario

````python
def stepseq(n0, n1, n2):
    """Genera u[n-n0] para n1 <= n <= n2. Devuelve (x, n)."""
    n = np.arange(n1, n2 + 1)
    x = (n >= n0).astype(float)
    return x, n

x_step, n_step = stepseq(0, -3, 3)
print("n =", n_step)
print("u[n] =", x_step)
````

### q) Rampa

````python
def rampseq(n0, n1, n2, m=1):
    """Genera una rampa de pendiente m: 0 si n<n0, m(n-n0) si n>=n0."""
    n = np.arange(n1, n2 + 1)
    x = m * np.maximum(n - n0, 0)
    return x.astype(float), n

x_ramp, n_ramp = rampseq(0, -3, 4, m=2)
print("n =", n_ramp)
print("r[n] =", x_ramp)
````

Estas funciones me ayudan a trabajar con desplazamientos de señales sin perder de vista en qué valor de n está cada muestra.

````python
def sigshift(x, n, k):
    """Devuelve la representación de x[n-k]: el eje de muestras pasa de n a n+k."""
    return np.asarray(x), np.asarray(n) + k


def align_signal(x, n, n_target):
    """Alinea (x,n) sobre n_target y rellena con ceros fuera del soporte."""
    x = np.asarray(x)
    n = np.asarray(n)
    n_target = np.asarray(n_target)
    y = np.zeros(n_target.shape, dtype=float)
    pos = {int(k): i for i, k in enumerate(n_target)}
    for value, k in zip(x, n):
        if int(k) in pos:
            y[pos[int(k)]] = value
    return y


def sigadd(x1, n1, x2, n2):
    """Suma dos secuencias alineándolas sobre la unión de sus índices."""
    n1 = np.asarray(n1)
    n2 = np.asarray(n2)
    n = np.arange(min(n1.min(), n2.min()), max(n1.max(), n2.max()) + 1)
    y = align_signal(x1, n1, n) + align_signal(x2, n2, n)
    return y, n
````

## 4. Punto r): Secuencias

En esta expresión hay dos impulsos en n=0, por lo que sus valores se suman: 3δ[n] + δ[n] = 4δ[n].

````python
n1 = np.arange(-5, 6)
x1 = np.zeros_like(n1, dtype=float)

for n0, coef in [(-1, 3), (-3, 5), (-2, 3), (0, 3), (0, 1)]:
    xi, _ = impseq(n0, n1[0], n1[-1])
    x1 += coef * xi

print("n =", n1)
print("x₁[n] =", x1)

plt.figure(figsize=(8, 3.5))
plt.stem(n1, x1, basefmt=' ')
plt.xlabel('n')
plt.ylabel('x₁[n]')
plt.title('x₁[n]')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

Un adelanto de 4 muestras se representa como u[n+4] y uno de 2 como u[n+2].

````python
n2 = np.arange(-6, 7)
u_n4, _ = stepseq(-4, -6, 6)  # u[n+4]
u_n2, _ = stepseq(-2, -6, 6)  # u[n+2]
x2 = -2 * u_n4 + 3 * u_n2

print("n =", n2)
print("x₂[n] =", x2)

plt.figure(figsize=(8, 3.5))
plt.stem(n2, x2, basefmt=' ')
plt.xlabel('n')
plt.ylabel('x₂[n]')
plt.title('x₂[n]')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

## 5. Punto s)  Secuencia por tramos y derivada discreta

En este punto se construye una señal por tramos y luego se calcula su diferencia entre muestras para aproximar la derivada discreta.

Cada tramo se genera con la función de rampa y luego se concatena. La derivada discreta se aproxima mediante la diferencia hacia atrás:

\[
\Delta x[n]=x[n]-x[n-1].
\]

````python
seg1, ns1 = rampseq(0, 0, 5)
seg2, ns2 = rampseq(5, 6, 11)
seg3, ns3 = rampseq(10, 12, 17)

x_s = np.concatenate([seg1, seg2, seg3])
n_s = np.concatenate([ns1, ns2, ns3])

# En n=0 no existe una muestra anterior dentro del intervalo.
# Se toma x[-1]=0 como condición de referencia, que coincide aquí con x[0]=0.
dx_s = np.diff(x_s, prepend=0)

print("n =", n_s)
print("x[n] =", x_s)
print("Δx[n] =", dx_s)
````

````python
fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)

ax[0].stem(n_s, x_s, basefmt=' ')
ax[0].set_ylabel('x[n]')
ax[0].set_title('Secuencia original')
ax[0].grid(alpha=0.3)

ax[1].stem(n_s, dx_s, linefmt='r', markerfmt='ro', basefmt=' ')
ax[1].set_xlabel('n')
ax[1].set_ylabel('Δx[n]')
ax[1].set_title('Diferencia hacia atrás')
ax[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
````

**Interpretación:** dentro de cada rampa la señal aumenta 1 por muestra. En los cambios de tramo aparecen saltos de -4. Por eso la diferencia discreta no coincide exactamente con la derivada de una señal continua.

### d-e) Calcular y agregar BMI

La guía pide calcular el BMI usando el peso en kg y la altura en metros. Primero se aplica la fórmula tal como aparece en el enunciado.

````python
df['BMI'] = df['Weight'] / (df['Height'] / 100) ** 2

display(df[['Weight', 'Height', 'BMI']].head(10))
````

### f) Categorizar el BMI solicitado
"
La guía propone cuatro categorías: **Bajo peso** "<18.5", **Normal** "18.5–24.9", **Sobrepeso** "25–29.9" y **Obesidad** "≥30".

Al aplicar la fórmula directamente a los datos del archivo, los valores resultan muy altos. Esto hace pensar que algunas unidades del archivo no coinciden con las unidades que necesita la fórmula.

````python
def categoria_imc(imc):
    if imc < 18.5:
        return 'Bajo peso'
    elif imc < 25:
        return 'Normal'
    elif imc < 30:
        return 'Sobrepeso'
    else:
        return 'Obesidad'


df['BMI_Category'] = df['BMI'].apply(categoria_imc)

display(df[['Weight', 'Height', 'BMI', 'BMI_Category']].head(10))
print("\nConteo por categoría:")
print(df['BMI_Category'].value_counts())
````

### Visualización de la distribución de "BMI_Category"

````python
orden = ['Bajo peso', 'Normal', 'Sobrepeso', 'Obesidad']
conteo = df['BMI_Category'].value_counts().reindex(orden, fill_value=0)

plt.figure(figsize=(6, 4))
conteo.plot(kind='bar', color='C0', edgecolor='black')
plt.xlabel('Categoría de BMI')
plt.ylabel('Número de personas')
plt.title('Distribución del BMI calculado con los datos originales')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
````

### Aclaración sobre las unidades del BMI

Al revisar los valores del archivo, se observa que "Weight" y "Height" parecen estar en unidades diferentes a las que pide la fórmula del BMI. Por ejemplo, valores como 118 para el peso y 64.5 para la altura son más coherentes con **libras** y **pulgadas** que con kg y cm.

Por eso, se hace la conversión a kg y metros y se vuelve a calcular el BMI. Después también se vuelven a aplicar las categorías para obtener un resultado acorde con las unidades correctas.

````python
# Conversión: libras -> kg y pulgadas -> metros
df['Weight_kg'] = df['Weight'] * 0.45359237
df['Height_m'] = df['Height'] * 0.0254

# Recalcular el BMI con las unidades convertidas
df['BMI_Correcto'] = df['Weight_kg'] / df['Height_m'] ** 2
df['BMI_Category_Correcto'] = df['BMI_Correcto'].apply(categoria_imc)

display(df[['Weight', 'Height', 'BMI', 'BMI_Correcto', 'BMI_Category_Correcto']].head(10).round({'BMI': 2, 'BMI_Correcto': 2}))

print("Conteo por categoría después de la conversión:")
print(df['BMI_Category_Correcto'].value_counts().reindex(orden, fill_value=0))
````

### Resultado final
Al hacer la conversión de unidades, los valores del BMI cambian a un rango mucho más razonable. Por eso, para interpretar los datos del archivo, se debe usar el BMI calculado después de convertir el peso y la altura a kg y metros.
Cabe destacar que al intentar calcular ek indice de masa corporal todos los sujetos se encontraban en obesidad, pero debido a que los datos suministardos estaban trucados con fines academicos, ya que donde debian estar los pesos se colocaron las alturas y viceversa. Despues de hacer el analisis, nuestra propuesta fue convertir las unidades del archivo original, que suponiamos estaban en libras y pulgadas, a las correctas para poder realizar el calculo y obtener valores dentro del rango sumistrado por la guía.
